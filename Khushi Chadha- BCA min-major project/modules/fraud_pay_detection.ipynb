```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.linear_model import LogisticRegression
import pickle
```

```python
data = pd.read_csv("C://Users//lenovo 2020//fruad_pay_detection.csv")
data.head()
```

```python
data= data[["amount","oldbalanceOrg","isFraud"]]
```

```python
data.info()
```

```python
X = data.drop("isFraud", axis=1)
y = data["isFraud"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=10)
```

```python
Log_reg=LogisticRegression(C=150, max_iter=150, class_weight='balanced')
Log_reg.fit(X_train, y_train)
log_pred = Log_reg.predict(X_test)

print(f'Accuracy score : {accuracy_score(log_pred, y_test)}')
print(f'Confusion matrix :\n {confusion_matrix(log_pred, y_test)}')
print(f'Classification report :\n {classification_report(log_pred, y_test)}')
```
