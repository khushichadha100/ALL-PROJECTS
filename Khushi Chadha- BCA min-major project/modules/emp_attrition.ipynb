```python
# Importing the necessary libraries
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split ,  GridSearchCV
from sklearn.metrics import accuracy_score, classification_report
import pickle
```

```python
# Importing the csv file
data = pd.read_csv("C://Users//lenovo 2020//Employee_performance.csv")
```

```python
data.head()
```

```python
for col in data.columns:
    unique_vals = data[col].unique().tolist()
    print(f"{col} = {unique_vals}")
```

```python
data.drop(['EmpNumber'],inplace=True,axis=1)
```

```python
# Here we have selected only the important columns
y = data.Attrition
#X = data.iloc[:,0:-1]  All predictors were selected it resulted in dropping of accuracy.
X = data.iloc[:,[9,16,20,21,22,23,24]] # Taking only variables with correlation coeffecient greater than 0.1
X.head()
```

```python
# Splitting into train and test for calculating the accuracy
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=10)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
# Standardization technique is used
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
# Training the model
from sklearn.ensemble import RandomForestClassifier

classifier_rfg=RandomForestClassifier(random_state=33,n_estimators=23)
parameters=[{'min_samples_split':[2,3,4,5],'criterion':['gini','entropy'],'min_samples_leaf':[1,2,3]}]

model_gridrf=GridSearchCV(estimator=classifier_rfg, param_grid=parameters, scoring='accuracy',cv=10)
model_gridrf.fit(X_train,y_train)
```

```python
model_gridrf.best_params_
```

```python
y_predict_rf = model_gridrf.predict(X_test)
```

```python
# Finding accuracy, precision, recall and confusion matrix
print(accuracy_score(y_test,y_predict_rf))
print(classification_report(y_test,y_predict_rf))
```
