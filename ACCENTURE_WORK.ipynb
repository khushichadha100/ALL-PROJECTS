IMPORTING NECESSARY LIBRARIES

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

HANDLING - 1ST FILE - CONTENT.CSV

```python
content=pd.read_csv('Content.csv')
```

```python
content.head()
```

DATA SHAPE AND OVERVIEW

```python
content.shape
```

```python
content.info()
```

REMOVING UN USEFUL COLUMNS

```python
content.drop(columns='URL',inplace=True)
content.drop(columns='User ID',inplace=True)
```

CHECKING NULL VALUES

```python
content.isnull().sum()
```

DATA CLEANING AND REPLACEMENT

```python
content['Category'].value_counts()
```

REPLACING QUOTATIONS FROM DATA

```python
content['Category']=content['Category'].str.replace('"',' ',)
```

CLEANED VALUES

```python
content['Category'].value_counts()
```

CHECKING DATA VALUES INSIDE OTHER COLUMNS ALSO

```python
content['Type'].value_counts()
```

```python
content['Content ID'].value_counts()
```

RENAMING COLUMN FOR BETTER UNDERSTANDABILITY

```python
content.rename(columns={'Type':'Content Type'},inplace=True)
```

CONTENT- FILE NOW READY FOR FURTHER PROCESSING

```python
content.info()
```

HANDLING - 2ND FILE - REACTIONS.CSV

```python
reactions=pd.read_csv('Reactions.csv')
```

```python
reactions.head()
```

DATA SHAPE AND OVERVIEW

```python
reactions.shape
```

```python
reactions.info()
```

RENAMING COLUMN FOR BETTER CLEARITY

```python
reactions.rename(columns={'Type':'Reaction Type'},inplace=True)
```

CHECKING AND HANDLING NULL VALUES

```python
reactions.isnull().sum()
```

DROPPING UN USEFUL COLUMNS AND ALSO NULL VALUES

```python
reactions.drop(columns='User ID',inplace=True)
```

```python
reactions.dropna(inplace=True)
```

```python
reactions.isnull().sum().sum()
```

```python
reactions.shape
```

DATA READY FOR FURTHER ANALYSIS

```python
reactions.info()
```

HANDLING - 3RD FILE - REACTIONS TYPE.CSV

```python
react_type=pd.read_csv('ReactionTypes.csv')
```

DATA OVERVIEW

```python
react_type.head()
```

CHECKING FOR NULL VALUES AND SHAPE OF THE DATA

```python
react_type.isnull().sum()
```

```python
react_type.shape
```

RENAMING COLUMNS FOR BETTER UNDERSTANDABILITY

```python
react_type.rename(columns={'Type':'Reaction Type'},inplace=True)
```

DATA IS READY FOR FURTHER PROCEDURE

```python
react_type.info()
```

MERGING ALL THE TABLES/DATAFRAMES TOGETHER FOR FURTHER ANALYSIS

```python
final_dataset=pd.concat([content,reactions,react_type],axis=1)
```

```python
final_dataset.head()
```

DROPPING DUPLICATE COLUMNS AS TABLES ARE INTERRELATED

```python
final_dataset = final_dataset.loc[:,~final_dataset.columns.duplicated()]
```

FINAL - CLEANED DATASET

```python
final_dataset.head()
```

TOP 5 CATEGORIES

```python
category_wise_scores = final_dataset.groupby('Category')['Score'].sum()
top_5_categories = category_wise_scores.sort_values(ascending=False).head(5)
top_5_categories= top_5_categories.reset_index()
top_5_categories.columns = ['Category', 'Total Score']
top_5_categories
```

TOP 5 CATEGORIES IN VISUALIZED FORMAT

```python
plt.figure(figsize=(8,4))
plt.barh(top_5_categories['Category'],top_5_categories['Total Score'],color='pink')
plt.gca().invert_yaxis() 
plt.show()
```
